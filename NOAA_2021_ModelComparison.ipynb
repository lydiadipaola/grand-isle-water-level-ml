```python
Grand Isle, Louisiana: Short-Term Water-Level Prediction

Objective: Predict one-hour-ahead coastal water levels at NOAA station 8761724 (Grand Isle, Louisiana) using historical water-level observations.

Data: NOAA hourly water-level observations, 2021.

Predictors: Water levels lagged by 1, 2, 3, 6, 12, and 24 hours.

Models: Linear Regression, Random Forest, and XGBoost.

Target: Water level one hour ahead.

Validation: Chronological 80/20 train-test split.

Evaluation: Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and R².

Overview: This project evaluates whether machine-learning and statistical models can use recent water-level observations to predict short-term coastal water-level variability.
```

```python
import pandas as pd
import matplotlib.pyplot as plt

comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "XGBoost"
    ],
    "MAE (m)": [0.022, 0.023, 0.022],
    "RMSE (m)": [0.030, 0.032, 0.030],
    "R²": [0.968, 0.966, 0.968]
})

display(comparison)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(comparison["Model"], comparison["MAE (m)"])

plt.ylabel("MAE (m)")
plt.title("Model Comparison: Mean Absolute Error")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(comparison["Model"], comparison["RMSE (m)"])

plt.ylabel("RMSE (m)")
plt.title("Model Comparison: Root Mean Squared Error")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(comparison["Model"], comparison["R²"])

plt.ylabel("R²")
plt.title("Model Comparison: R²")

plt.tight_layout()
plt.show()
```

All three models demonstrated strong performance for one-hour-ahead water-level prediction at Grand Isle. Linear Regression and XGBoost achieved nearly identical performance, while Random Forest produced slightly higher prediction errors. The similar performance across models suggests that recent water-level observations contain substantial information about short-term water-level variability. The strong performance of the linear baseline also indicates that increasing model complexity did not substantially improve predictive accuracy for this dataset.
